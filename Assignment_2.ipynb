```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

model.eval()
```

```python
def generate_text(prompt, temperature=1.0, top_k=0, top_p=1.0,
                  repetition_penalty=1.0):

    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=100,
            do_sample=True,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(output[0], skip_special_tokens=True)


prompt = "The young scientist entered the laboratory and discovered"

# Experiment 1
output1 = generate_text(
    prompt,
    temperature=0.2
)

# Experiment 2
output2 = generate_text(
    prompt,
    temperature=1.0
)

# Experiment 3
output3 = generate_text(
    prompt,
    top_k=20
)

# Experiment 4
output4 = generate_text(
    prompt,
    top_p=0.8
)

# Experiment 5
output5 = generate_text(
    prompt,
    repetition_penalty=1.5
)

print("Experiment 1:")
print(output1)

print("\nExperiment 2:")
print(output2)

print("\nExperiment 3:")
print(output3)

print("\nExperiment 4:")
print(output4)

print("\nExperiment 5:")
print(output5)
```

```python
custom_prompt = "Riya opened the mysterious box and found a small glowing key."

for i in range(3):
    result = generate_text(
        custom_prompt,
        temperature=0.7 + (i * 0.3),
        top_p=0.9
    )

    print(f"Version {i+1}:")
    print(result)
    print()
```
